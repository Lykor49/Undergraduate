### 1. Tensor 张量
- 定义: PyTorch 中用于存储和计算多维数据的基本数据结构，支持自动微分和 GPU 加速
- 原理: 按照维度数量, Tensor可以表示为 Scalar标量 (0D)、Vector向量 (1D)、Matrix矩阵 (2D)、高维张量 (ND)


```python
import torch

# 语法 
# torch.tensor(data)  # 根据给定数据创建 Tensor

# 示例
a = torch.tensor(5)                 # 创建标量（0D）
b = torch.tensor([1, 2, 3])         # 创建向量（1D）
c = torch.tensor([[1, 2], [3, 4]])  # 创建矩阵（2D）

print(a)  # tensor(5)
print(b)  # tensor([1, 2, 3])
print(c)  # 输出二维张量

# 测试
# ex. 创建矩阵 d = [[1,2,3],[4,5,6]]
d = torch.tensor([[1,2,3], [4,5,6]])
print(d)
```

### 2. Shape / ndim / dtype / device 张量属性
- Shape (形状): 表示张量各维度的大小
- ndim (维度): 表示张量的维度数量
- dtype (数据类型): 表示张量中元素的数据类型
- device (计算设备): 表示张量所在的计算设备

```python
import torch  # 导入 PyTorch

# 语法
x = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)  # 创建二维浮点张量

print(x.shape)   # 获取张量形状, torch.Size([2, 3])
print(x.ndim)    # 获取维度数量, 2
print(x.dtype)   # 获取数据类型, torch.float32
print(x.device)  # 获取计算设备, cpu

print(x.shape[0])  # 第一维大小：2
print(x.shape[1])  # 第二维大小：3

y = x.to(torch.float64)  # 转换为 64 位浮点数
print(y.dtype)           # torch.float64

# 常见的数据类型与计算设备
# dtype：torch.float32、torch.float64、torch.int64、torch.bool
# device：cpu（处理器）、cuda（NVIDIA GPU）、mps（Apple GPU）

# 测试
# 创建一个形状为 (3, 4) 的随机 Tensor：
# 1. 输出 shape、ndim、dtype、device。
# 2. 获取第一维和第二维的大小。
# 3. 将其数据类型转换为 torch.float64

a = torch.randn(3, 4)

print(a.shape)
print(a.ndim)
print(a.dtype)
print(a.device)

print(a.shape[0])
print(a.shape[1])

b = a.to(torch.float64)
print(b.dtype)
```

### 3. Indexing / Slicing 索引与切片
- Indexing (索引): 通过索引位置访问 Tensor 中的指定元素
- Slicing (切片): 通过索引范围提取 Tensor 中的部分数据

切片基本原则

`x[start:stop:step]`

- start：起始索引 (包含)
- stop：结束索引 (不包含)
- step：步长，默认为 1


```python
import torch  # 导入 PyTorch

x = torch.tensor([[1, 2, 3], [4, 5, 6]])  # 创建二维 Tensor

# Indexing 索引
print(x[0, 0])    # 第一行第一列：1
print(x[1, 2])    # 第二行第三列：6
print(x[-1])      # 最后一行：[4, 5, 6]

# Slicing 切片
print(x[0, :])    # 第一行：[1, 2, 3]
print(x[:, 1])    # 第二列：[2, 5]
print(x[:, 0:2])  # 前两列
print(x[:, ::2])  # 每隔一列提取一次
```

### 4. Basic Operations 基础运算
- 定义: PyTorch 支持对 Tensor 进行加、减、乘、除及矩阵运算等数值计算

```python
import torch  # 导入 PyTorch

a = torch.tensor([[1., 2.], [3., 4.]])  # 创建矩阵 A    .表示浮点型张量
b = torch.tensor([[5., 6.], [7., 8.]])  # 创建矩阵 B

# 基本算术运算
print(a + b)   # 逐元素加法
print(a - b)   # 逐元素减法
print(a * b)   # 逐元素乘法
print(a / b)   # 逐元素除法
print(a ** 2)  # 逐元素平方

# 矩阵运算
print(a @ b)   # 矩阵乘法

# 统计运算
print(torch.sum(a))   # 所有元素之和：10
print(torch.mean(a))  # 所有元素平均值：2.5
```
