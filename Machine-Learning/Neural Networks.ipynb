# 单层前向传播NumPy实现

```python
import numpy as np

# Example parameters

W = np.array([
    [1, -3,  5],
    [2,  4, -6]
])
b = np.array([-1, 1, 2])
a_in = np.array([-2, 4])

def dense(a_in, W, b, g):
    units = W.shape[1]
    a_out = np.zeros(units)

    for j in range(units):
        w = W[:, j]
        z = np.dot(w, a_in) + b[j]
        a_out[j] = g(z)

    return a_out

def sequential(x):
    a1 = dense(x, W1, b1)
    a2 = dense(a1, W2, b2)
    a3 = dense(a2, W3, b3)
    a4 = dense(a3, W4, b4)

    f_x = a4

    return f_x
```

## activation function 激活函数
### Sigmoid

$$
g(z)=\frac{1}{1+e^{-z}}
$$

### ReLU

$$
g(z)=\max(0,z)
$$

or equivalently,

$$
g(z)=
\begin{cases}
z, & z>0 \\
0, & z\leq 0
\end{cases}
$$

### Linear

$$
g(z)=z
$$

## Softmax定义式
$$
z_j
=
\mathbf{w}_j \cdot \mathbf{x}
+
b_j,
\quad
j=1,\ldots,N
$$

$$
a_j
=
\frac{e^{z_j}}
{\sum_{k=1}^{N} e^{z_k}}
=
P(y=j\mid\mathbf{x})
$$

## Crossentropy loss 交叉熵损失
$$
\mathrm{loss}(a_1,\ldots,a_N,y)
=
\begin{cases}
-\log a_1, & \text{if } y=1 \\
-\log a_2, & \text{if } y=2 \\
\vdots & \\
-\log a_N, & \text{if } y=N
\end{cases}
$$

## Training set / Cross validation set / Test set

### Training Error

$$
J_{\mathrm{train}}(\mathbf{w}, b)
=
\frac{1}{2m_{\mathrm{train}}}
\sum_{i=1}^{m_{\mathrm{train}}}
\left(
f_{\mathbf{w},b}\left(\mathbf{x}^{(i)}\right)
-
y^{(i)}
\right)^2
$$

### Cross Validation Error

$$
J_{\mathrm{cv}}(\mathbf{w}, b)
=
\frac{1}{2m_{\mathrm{cv}}}
\sum_{i=1}^{m_{\mathrm{cv}}}
\left(
f_{\mathbf{w},b}\left(\mathbf{x}_{\mathrm{cv}}^{(i)}\right)
-
y_{\mathrm{cv}}^{(i)}
\right)^2
$$

Cross validation error is also called **validation error** or **dev error**.

### Test Error

$$
J_{\mathrm{test}}(\mathbf{w}, b)
=
\frac{1}{2m_{\mathrm{test}}}
\sum_{i=1}^{m_{\mathrm{test}}}
\left(
f_{\mathbf{w},b}\left(\mathbf{x}_{\mathrm{test}}^{(i)}\right)
-
y_{\mathrm{test}}^{(i)}
\right)^2
$$
