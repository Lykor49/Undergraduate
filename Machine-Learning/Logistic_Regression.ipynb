# Logistic_Regression 逻辑回归
## Logistic_Regression定义式

$$
f_{\mathbf{w},b}(\mathbf{x}) = g(\mathbf{w}\cdot\mathbf{x}+b)
$$

where $g(z)$ is the **Sigmoid Function**:

$$
g(z)=\frac{1}{1+e^{-z}}
$$

## Logistic loss function
- 逻辑回归的代价函数

$$
J(\mathbf{w},b)
=
\frac{1}{m}
\sum_{i=1}^{m}
L\left(
f_{\mathbf{w},b}\left(\mathbf{x}^{(i)}\right),
y^{(i)}
\right)
$$

- 逻辑回归的损失函数

$$
L\left(f_{\mathbf{w},b}(\mathbf{x}),y\right)
=
\begin{cases}
-\log\left(f_{\mathbf{w},b}(\mathbf{x})\right), & y=1 \\
-\log\left(1-f_{\mathbf{w},b}(\mathbf{x})\right), & y=0
\end{cases}
$$

- 简化

$$
L\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)}),y^{(i)}\right)
=
-y^{(i)}\log\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)
-
\left(1-y^{(i)}\right)
\log\left(1-f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)
$$

## Gradient Descent for Logistic Regression 梯度下降
Repeat until convergence:
$$
\left\{
\begin{aligned}
w_j
&=
w_j
-
\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
\left(
f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}
\right)
x_j^{(i)}
\right]
\\[8pt]
b
&=
b
-
\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
\left(
f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}
\right)
\right]
\end{aligned}
\right.
$$


- Linear Regression
$
f_{\mathbf{w},b}(\mathbf{x})
=
\mathbf{w}\cdot\mathbf{x}+b
$

- Logistic Regression
$
f_{\mathbf{w},b}(\mathbf{x})
=
\frac{1}
{1+e^{-(\mathbf{w}\cdot\mathbf{x}+b)}}
$

## Cost Function with Regularization 正则化的代价函数

- Cost Function of linear regression
 
$$
J(\mathbf{w},b)
=
\frac{1}{2m}
\sum_{i=1}^{m}
\left(
f_{\mathbf{w},b}\left(\mathbf{x}^{(i)}\right)-y^{(i)}
\right)^2
+
\frac{\lambda}{2m}
\sum_{j=1}^{n}
w_j^2
$$

- Gradient descent of Regularized linear regression

$$
\text{repeat}
\left\{
\begin{aligned}
w_j
&=
w_j
-
\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
\left(
f_{\mathbf{w},b}\left(\mathbf{x}^{(i)}\right)-y^{(i)}
\right)
x_j^{(i)}
+
\frac{\lambda}{m}w_j
\right]
\\[8pt]
b
&=
b
-
\alpha
\frac{1}{m}
\sum_{i=1}^{m}
\left(
f_{\mathbf{w},b}\left(\mathbf{x}^{(i)}\right)-y^{(i)}
\right)
\end{aligned}
\right.
$$

- Gradient descent of Regularized logistic regression
- 同Linear regression, 代价函数不同, 梯度下降导数推导相同
