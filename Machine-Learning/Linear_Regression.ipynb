# Linear Regression 线性回归

- Linear Regression 线性回归定义式

$$
\hat{y}=f_{w,b}(x)=wx+b
$$

$\hat{y}$ 为预测值，$w$ 为权重，$b$ 为偏置

- Cost Function 代价函数定义式

$$
J(w,b)=\frac{1}{2m}\sum_{i=1}^{m}\left(f_{w,b}\left(x^{(i)}\right)-y^{(i)}\right)^2
$$

$m$ = number of training examples

- Gradient descent algorithm 梯度下降算法

$$
\text{repeat until convergence:}
$$
$$
w = w - \alpha \frac{\partial}{\partial w} J(w,b)
$$
$$
b = b - \alpha \frac{\partial}{\partial b} J(w,b)
$$

$\alpha$ 为学习率 (Learning Rate)

- Multiple Linear Regression 多元线性回归
$$
f_{\vec{w},b}(\vec{x})
=
w_1x_1+w_2x_2+\cdots+w_nx_n+b
=
\vec{w}\cdot\vec{x}+b
$$
$$
\vec{w} = [w_1,\ w_2,\ \cdots,\ w_n]
$$
$$
\vec{x} = [x_1,\ x_2,\ \cdots,\ x_n]
$$
$$
b \in \mathbb{R}
$$

- 多元线性回归 Numpy实现

```python
import numpy as np

w = np.array([1.0, 2.5, -3.3])
b = 4
x = np.array([10, 20, 30])

# Without vectorization
f = w[0] * x[0] + w[1] * x[1] + w[2] * x[2] + b 

# Vectorization
f = np.dot(w,x) + b
```
