简单MLP

```python
import torch
from torch import nn

# 1. 数据
x = torch.tensor([[1.0], [2.0], [3.0]])
y = torch.tensor([[2.0], [4.0], [6.0]])

# 2. MLP模型
model = nn.Sequential(
    nn.Linear(1, 8),
    nn.ReLU(),
    nn.Linear(8, 1)
)

# 3. 损失函数与优化器
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# 4. 训练
for epoch in range(1000):
    y_pred = model(x)             # Forward：前向传播
    loss = loss_fn(y_pred, y)     # Loss：计算损失
    optimizer.zero_grad()         # 清空梯度
    loss.backward()               # Backward：计算梯度
    optimizer.step()              # Optimizer：更新参数

print(model(torch.tensor([[4.0]])).item())  # 预测结果，通常接近8
```

Autograd 自动求导

```python
requires_grad=True	# 让 PyTorch 跟踪相关运算，以便求导
loss.backward()	    # 反向传播，计算梯度
.grad	            # 查看张量中保存的梯度
```

```python
import torch  # 导入 PyTorch

x = torch.tensor(2.0)  # 输入数据
w = torch.tensor(3.0, requires_grad=True)  # 需要求梯度的参数

y = w * x**2  # 前向传播，y=12
y.backward()  # 反向传播，计算梯度

print(y.item())  # 输出：12.0
print(w.grad.item())  # 输出：4.0
```

Gradient Control（梯度控制）

```python
requires_grad_(False)	# 不再计算该参数自身的梯度
detach()	            # 生成一个脱离当前计算图的张量
torch.no_grad()	        # 在指定代码块中关闭梯度记录
```

```python
import torch  # 导入PyTorch

# 1. 冻结参数
x = torch.tensor(2.0, requires_grad=True)   # 需要计算x的梯度
w = torch.tensor(3.0, requires_grad=False)  # 冻结w
y = w * x       # 前向计算
y.backward()    # 反向传播
print(x.grad)   # tensor(3.)
print(w.grad)   # None

# 2. detach：截断计算图
a = torch.tensor(2.0, requires_grad=True)  # 创建可求导张量
b = a * 3                # 建立计算图
c = b.detach()           # 截断与a的梯度连接
print(c.requires_grad)   # False

# 3. no_grad：关闭梯度记录
with torch.no_grad():    # 进入不记录梯度的环境
    z = a * 3            # 不建立求导计算图
print(z.requires_grad)   # False
```
