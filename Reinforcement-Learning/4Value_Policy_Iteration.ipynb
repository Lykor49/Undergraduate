## Lecture 4: Value / Policy Iteration 值迭代 / 策略迭代

1. Value Iteration 值迭代

核心过程:

$$
v_0
\rightarrow
q_0
\rightarrow
\text{Greedy}
\rightarrow
\pi_1
\rightarrow
v_1
\rightarrow
\cdots
\rightarrow
v^*
$$

核心公式: 

$$
q_k(s,a)
=
\sum_r p(r\mid s,a)r
+
\gamma\sum_{s'}p(s'\mid s,a)v_k(s')
$$

$$
a_k^*(s)
=
\arg\max_a q_k(s,a)
$$

$$
v_{k+1}(s)
=
\max_a q_k(s,a)
$$

2. Truncated Policy Iteration 截断策略迭代

核心过程: 

$$
\pi_0
\rightarrow
v_{\pi_0}^{(0)}
\rightarrow
v_{\pi_0}^{(1)}
\rightarrow
\cdots
\rightarrow
v_{\pi_0}^{(J)}
\rightarrow
q_0
\rightarrow
\text{Greedy}
\rightarrow
\pi_1
$$

核心公式: 

$$
v_{\pi_0}^{(j+1)}
=
r_{\pi_0}
+
\gamma P_{\pi_0}v_{\pi_0}^{(j)}
$$

做到第J次停止:

$$
v_{\pi_0}^{(J)}
\rightarrow
q_0
\rightarrow
\text{Greedy}
\rightarrow
\pi_1
$$

3. Policy Iteration 策略迭代

核心过程:

$$
\pi_0
\rightarrow
v_{\pi_0}^{(0)}
\rightarrow
v_{\pi_0}^{(1)}
\rightarrow
\cdots
\rightarrow
v_{\pi_0}
\rightarrow
q_{\pi_0}
\rightarrow
\text{Greedy}
\rightarrow
\pi_1
$$

Policy Evaluation

$$
v_{\pi_0}^{(j+1)}
=
r_{\pi_0}
+
\gamma P_{\pi_0}v_{\pi_0}^{(j)}
$$

一直迭代到:

$$
v_{\pi_0}^{(j)}
\rightarrow
v_{\pi_0}
$$

Policy Improvement

$$
a_0^*(s)
=
\arg\max_a q_{\pi_0}(s,a)
$$

迭代: 

$$
\pi_0
\rightarrow
v_{\pi_0}
\rightarrow
q_{\pi_0}
\rightarrow
\text{Greedy}
\rightarrow
\pi_1
$$
