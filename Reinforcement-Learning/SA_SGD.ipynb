## Lecture 6: SA / SGD

### Mean estimation 均值估计

1. 引入: Monte Carlo Estimation 蒙特卡洛估计

$$
\mathbb{E}[X]
\approx
\frac{1}{n}
\sum_{i=1}^{n} x_i
$$

- 作用：用大量随机样本的平均值近似真实期望

2. 引出: 增量式更新 (incremental)

$$
w_{k+1}
=
w_k
+
\alpha_k(x_k-w_k)
$$

- 原因: 避免收集大量样本后再统一求平均，引出增量式更新
- 作用: 每获得一个新样本，可以在旧估计的基础上立即修正得到新的估计


### Stochastic Approximation 随机近似

- 随机近似: 指的是一大类用于求解方程或优化问题的随机迭代算法

### Robbins-Monro Algorithm 罗宾斯-门罗算法 (RM)
1. 定义: 一种随机近似算法，通过不断利用目标函数的带噪声观测进行迭代，从而逼近方程 \(g(w)=0\) 的真实根

- 观测值
$$
\tilde{g}(w,\eta)
=
g(w)+\eta
$$

- RM算法核心更新
$$
w_{k+1}
=
w_k
-
a_k\tilde{g}(w_k,\eta_k)
$$

- RM算法不需要关于函数表达式的任何信息, 仅需要输入和输出

2. Convergence 收敛性质

- **函数满足良好的单调性：**

$$
0<c_1\leq\nabla_w g(w)\leq c_2
$$

- **Step Size 满足：**

$$
\sum_{k=1}^{\infty}a_k=\infty,
\qquad
\sum_{k=1}^{\infty}a_k^2<\infty
$$

典型：

$$
a_k=\frac{1}{k}
$$

- **Noise 满足零均值、有限方差：**

$$
\mathbb{E}[\eta_k\mid H_k]=0
$$

$$
\mathbb{E}[\eta_k^2\mid H_k]<\infty
$$

- **收敛直觉：** 步长需要逐渐减小以抑制随机噪声，但又不能减小得太快，否则可能还没有到达真实根就停止有效更新。


### Stochastic Gradient Descent 随机梯度下降
随机梯度下降算法是一种特殊的RM算法, 而期望值估计算法是一种特殊的随机梯度下降算法

1. 解决问题: 当优化目标中包含期望，真实梯度很难直接计算时，用随机样本得到的梯度来近似真实梯度，并逐步更新参数

优化目标:
$$
\min_{w} J(w)
=
\mathbb{E}[f(w,X)]
$$


2. Gradient Descent 梯度下降 (GD)

$$
w_{k+1}
=
w_k
-
\alpha_k
\nabla_w J(w_k)
$$

3. Batch Gradient Descent 批量梯度下降 (BGD)

$$
w_{k+1}
=
w_k
-
\alpha_k
\frac{1}{n}
\sum_{i=1}^{n}
\nabla_w f(w_k,x_i)
$$

4. Stochastic Gradient Descent 随机梯度下降 (SGD)

$$
w_{k+1}
=
w_k
-
\alpha_k
\nabla_w f(w_k,x_k)
$$
