## Lecture 3: Bellman Optimality Equation 贝尔曼最优方程

1. Optimal Policy 最优策略 / Optimal State Value 最优状态价值

最优策略定义式：

$$
v_{\pi^*}(s) \geq v_\pi(s), \quad \forall s \in \mathcal{S}, \forall \pi
$$

最优状态价值定义式：

$$
v^*(s) = \max_{\pi} v_\pi(s) = v_{\pi^*}(s)
$$

问题：
- 存在性：最优策略存在吗？
- 唯一性：最优策略唯一吗？
- 随机性：最优策略是随机的还是确定的？
- 算法： 如何得到最优策略和最优状态值？

2. Bellman Optimality Equation 贝尔曼最优方程

展开形式：

$$
v(s)=
\max_{\pi}
\sum_a \pi(a\mid s)
\left(
\sum_r p(r\mid s,a)r
+
\gamma\sum_{s'}p(s'\mid s,a)v(s')
\right),
\quad \forall s\in\mathcal{S}
$$

Q Value 形式

$$
v(s)=
\max_{\pi}
\sum_a \pi(a\mid s)q(s,a),
\quad s\in\mathcal{S}
$$


3. Greedy Policy 贪婪策略

核心公式：选择最大 Q Value 的动作

$$
a^*(s)=\arg\max_a q(s,a)
$$

推导公式：策略最大化等价于动作价值最大化

$$
\max_\pi\sum_a\pi(a\mid s)q(s,a)
=
\max_a q(s,a)
$$

贪婪策略表达式：

$$
\pi_{\mathrm{greedy}}(a\mid s)=
\begin{cases}
1, & a=a^*(s)\\
0, & a\neq a^*(s)
\end{cases}
$$

4. 矩阵-向量形式

$$
v = \max_{\pi}\left(r_{\pi}+\gamma P_{\pi}v\right)
$$

5. Contraction Mapping Theorem 压缩映射定理

- Fixed Point 不动点

$$
f(x^*)=x^*
$$

- Contraction Mapping 压缩映射

$$
\|f(x_1)-f(x_2)\|
\leq
\gamma\|x_1-x_2\|,
\quad 0<\gamma<1
$$

- Contraction Mapping Theorem 压缩映射定理

$$
x_{k+1}=f(x_k)
$$

$$
\lim_{k\to\infty}x_k=x^*
$$

$$
f(x^*)=x^*
$$

6. Iterative Solution of BOE 贝尔曼最优方程的迭代求解

- 最优状态价值迭代

$$
v_{k+1}=f(v_k)
=\max_{\pi}\left(r_{\pi}+\gamma P_{\pi}v_k\right)
$$

- 迭代收敛

$$
\lim_{k\to\infty}v_k=v^*
$$

- 根据最优状态价值求解最优策略

$$
a^*(s)=\arg\max_a
\left[
\sum_r p(r\mid s,a)r
+\gamma\sum_{s'}p(s'\mid s,a)v^*(s')
\right]
$$
