## Lecture 2: Bellman Equation

1. State Value Function 状态价值函数
$$
v_\pi(s)=\mathbb{E}[G_t \mid S_t=s]
$$

2. Bellman Equation 贝尔曼公式

定义式：
$$
v_\pi(s)
=
\mathbb{E}_\pi
\left[
R_{t+1}
+
\gamma v_\pi(S_{t+1})
\mid
S_t=s
\right]
$$

展开形式：
$$
v_\pi(s)
=
\sum_a \pi(a|s)
\left[
\sum_r p(r|s,a)r
+
\gamma
\sum_{s'} p(s'|s,a)v_\pi(s')
\right]
$$
