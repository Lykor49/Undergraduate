## Lecture 2: Bellman Equation

1. State Value 状态价值
$$
v_\pi(s)=\mathbb{E}[G_t \mid S_t=s]
$$

2. Bellman Equation 贝尔曼公式

定义式：
$$
v_\pi(s)
=
\mathbb{E}_\pi
\left[
R_{t+1}
+
\gamma v_\pi(S_{t+1})
\mid
S_t=s
\right]
$$

展开形式：
$$
v_\pi(s)
=
\sum_a \pi(a|s)
\left[
\sum_r p(r|s,a)r
+
\gamma
\sum_{s'} p(s'|s,a)v_\pi(s')
\right]
$$

3. Matrix-vector form

核心公式: 
$$
v_\pi = r_\pi + \gamma P_\pi v_\pi
$$

解析解:
$$
v_\pi = (I - \gamma P_\pi)^{-1} r_\pi
$$

4. Action Value 动作价值

定义式:
$$
q_\pi(s,a)=\mathbb{E}_\pi[G_t\mid S_t=s,A_t=a]
$$

核心公式:
$$
q_\pi(s,a)
=
\sum_r p(r\mid s,a)r
+
\gamma\sum_{s'}p(s'\mid s,a)v_\pi(s')
$$

与State Value的关系: 
$$
v_\pi(s)=\sum_a\pi(a\mid s)q_\pi(s,a)
$$
