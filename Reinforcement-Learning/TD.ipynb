## Lecture 7: Temporal-Difference Learning 时序差分方法
- TD算法、Sarsa算法、n-Step Sarsa算法、Q-Learning算法

### Temporal-Difference Learning 时序差分算法

1. TD Learning (estimate state value)
- 定义: 给定策略, 利用当前一步获得的 Reward 和下一状态的 Value estimate 来更新当前状态的 Value

核心更新:
$$
v_{t+1}(s_t)
=
v_t(s_t)
-
\alpha_t(s_t)
\left[
v_t(s_t)
-
\left(
r_{t+1}
+
\gamma v_t(s_{t+1})
\right)
\right]
$$

$$
v_{t+1}(s)
=
v_t(s),
\qquad
\forall s \neq s_t
$$

2. TD target
- 定义: TD Target 是“即时奖励 + 下一状态估计价值”，作为当前状态应该逼近的目标

$$
\bar{v}_t
=
r_{t+1}
+
\gamma v_t(s_{t+1})
$$

3. TD error
- 定义: TD Error 衡量当前 Value estimate 与 TD Target 之间的误差

$$
\delta_t
=
v_t(s_t)
-
\left[
r_{t+1}
+
\gamma v_t(s_{t+1})
\right]
$$

### Sarsa 算法
1. Sarsa (estimate action value)

$$
q_{t+1}(s_t,a_t)
=
q_t(s_t,a_t)
-
\alpha_t(s_t,a_t)
\left[
q_t(s_t,a_t)
-
\left(
r_{t+1}
+
\gamma q_t(s_{t+1},a_{t+1})
\right)
\right]
$$

$$
q_{t+1}(s,a)
=
q_t(s,a),
\qquad
\forall (s,a)\neq(s_t,a_t)
$$
